# ****EXPLORATORY DATA ANALYSIS (EDA)****

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salary_df = pd.read_csv('../data/salary_cleaned.csv')
print("============DATASET============:\n")
display(salary_df.head())
```

```python
print("============SHAPE============:")
print("columns: ", salary_df.shape[1])
print("rows: ", salary_df.shape[0])
print("============INFO============:")
display(salary_df.info())
print("============DESCRIBE============:")
display(salary_df.describe(include='all'))
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(salary_df.select_dtypes(include=['int64', 'float64']).corr(), annot=True, fmt='.2f', linewidths=0.5, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
